## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# Clean and split a server record into a dictionary
raw = " Srv-01, used=87, total=100 "

#Remove whitespace from the whole line
raw = raw.strip()

#Separate the hostname from the remaining data
hostname_part, rest = raw.split(",", 1)

#Clean and standardise the hostname
hostname = hostname_part.strip().lower()

#Store the hostname and extract the remaining values
data = {"hostname": hostname}

for item in rest.split(","):          # Split the remaining data into individual pieces
    key, value = item.split("=")      #Separate each piece into a key and value
    data[key.strip()] = float(value.strip())   #Remove whitespace and convert the value to a float

print(data)
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# Create a reusable function to parse server records
def parse_line(raw_line):
    """Parse a server record and return its data as a dictionary."""
    raw_line = raw_line.strip()

    # Separate and clean the hostname
    hostname_part, rest = raw_line.split(",", 1)
    hostname = hostname_part.strip().lower()

    # Store the hostname and parse the remaining values
    data = {"hostname": hostname}
    for item in rest.split(","):
        key, value = item.split("=")
        data[key.strip()] = float(value.strip())

    return data
print(parse_line("SRV-02, used=120,total=100"))
````

## Challenge · for a first

*These combine several ideas from this week. There are no step-by-step instructions.*

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
# Read and process server records from file
def main():
    """Read server records and display their status."""
    #Open file and read it line by line
    with open("messy_lines.txt", "r") as f:
        for line in f:
            # Parse each line
            data = parse_line(line)
            #Calculate percentage of server usage
            percentage = (data["used"] / data["total"]) * 100
            # Determine server status
            if percentage >= 100:
                status = "OVER LIMIT"
            elif percentage >= 90:
                status = "WARNING"
            else:
                status = "OK"
            print(data["hostname"], status)
main()
````

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print

         "srv-08,used=unknown,total=100",
